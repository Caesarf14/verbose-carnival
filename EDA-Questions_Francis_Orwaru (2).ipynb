```python
#9.Find out the count of male and female penguins
gender_counts = df_cleaned['sex'].value_counts()
print("\nCount of Male and Female Penguins:\n",gender_counts)
```

```python
#10.Find out the count of species 
species_counts = df_cleaned['species'].value_counts()
print("\nCount of Each Species:\n", species_counts)
```

```python
#11. Use a countplot graph to display the number of male and female penguins
#Give the plot a title "Count of male and female penguins"
#Give a xlabel and ylabel 
def sexes_countplot():
    plt.figure(figsize=(8, 5))
    sns.countplot(x='sex', data=df)
    plt.title("Count of Male and Female Penguins")
    plt.xlabel("Sex")
    plt.ylabel("Count")
    plt.show()
sexes_countplot()
```

```python
#12. For all the penguins display the bill_length_mm and bill_depth_mm using pairplot
def penguins_pairplot():
    sns.pairplot(df_cleaned, vars=['bill_length_mm', 'bill_depth_mm'], hue='species')
    plt.show()
penguins_pairplot()
```

```python
#13.Create a visualization to display the flipper_length_mm and body_mass_g for the penguin dataset.
def penguin_scatterplot():
    plt.figure(figsize=(8, 5))
    sns.scatterplot(x='flipper_length_mm', y='body_mass_g', hue='species' if 'species' in df_cleaned.columns else None, data=df_cleaned)
    plt.title("Flipper Length vs Body Mass")
    plt.xlabel("Flipper Length (mm)")
    plt.ylabel("Body Mass (g)")
    plt.show()
penguin_scatterplot()
```

```python
#14. Create a histogram with the flipper_length_mm and bins = 10
def flipper_histogram():
    plt.figure(figsize=(8, 5))
    sns.histplot(df_cleaned['flipper_length_mm'], bins=10, kde=True)
    plt.title("Histogram of Flipper Length")
    plt.xlabel("Flipper Length (mm)")
    plt.ylabel("Frequency")
    plt.show()
flipper_histogram()
```

```python
#17.Check if there exists any outliers based on the boxplot based on numerical values  of penguins_cleaned dataset
for col in df_cleaned.select_dtypes(include=['float64', 'int64']).columns:
    plt.figure(figsize=(8, 5))
    sns.boxplot(x=col, data=df_cleaned)
    plt.title(f"Boxplot of {col}")
    plt.show()
```

```python
#15. Create a joint point with body_length_mm, body_depth_mm for penguins_cleaned dataset
def body_jointplot():
    sns.jointplot(x='bill_length_mm', y='bill_depth_mm', data=df_cleaned, kind='scatter', hue='species' if 'species' in df_cleaned.columns else None)
    plt.show()
body_jointplot()
```

```python
#16. Use a pairwise plot to display all the numerical values of the dataset penguins_cleaned
def numerical_pairplot():
    sns.pairplot(df_cleaned.select_dtypes(include=['float64', 'int64']))
    plt.show()
numerical_pairplot()
```

```python
#18. Create a violin plot to display the body_mass_g of different species based on sex
def body_violinplot():
    plt.figure(figsize=(8, 5))
    sns.violinplot(x='species', y='body_mass_g', hue='sex', data=df_cleaned, split=True)
    plt.title("Body Mass by Species and Sex")
    plt.xlabel("Species")
    plt.ylabel("Body Mass (g)")
    plt.show()
body_violinplot()
```
